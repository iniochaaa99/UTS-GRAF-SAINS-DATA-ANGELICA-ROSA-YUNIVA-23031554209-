```python
import pandas as pd

P1 = ("2015-05-28", "2015-08-16")   # training
P2 = ("2015-08-17", "2015-09-06")   # testing

df = pd.read_csv("data/events.csv")
df["dt"] = pd.to_datetime(df["timestamp"], unit="ms")   # milidetik, UTC
print("Rentang data:", df["dt"].min(), "->", df["dt"].max())


def cut(df, start, end):
    s = pd.Timestamp(start)
    e = pd.Timestamp(end) + pd.Timedelta(days=1)   # tanggal akhir ikut penuh
    return df[(df["dt"] >= s) & (df["dt"] < e)].drop(columns="dt")


p1, p2 = cut(df, *P1), cut(df, *P2)
p1.to_csv("data/events_p1.csv", index=False)
p2.to_csv("data/events_p2.csv", index=False)
print("Baris P1:", len(p1), "| Baris P2:", len(p2))
```
